```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(
    n_samples=100,           # Total number of data points
    n_features=2,          # Total number of dimensions
    n_informative=1,       # Number of features carrying signal
    n_redundant=0,         # No linear combinations of informative features
    n_classes=2,           # Binary classification
    n_clusters_per_class=1, # One cluster per label
    random_state=41,       # Reproducible results
    hypercube=False,       # Clusters not restricted to hypercube vertices
    class_sep=10           # High separation for distinct classes
)

print(f"X shape: {X.shape}") # Should be (100, 2)
print(f"y shape: {y.shape}") # Should be (100,)
```

```python
plt.figure(figsize = (10,6))
plt.scatter(X[:,0],X[:,1],cmap = "winter" ,c = y, s = 100)
```

```python
def perceptron(X,y):
    X = np.insert(X,0,1,axis = 1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(100):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j] , weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z >0 else 0    # z = x2b2+x1b1+b0
```

```python
intercept_ , coef_ = perceptron(X,y)
```

```python
intercept_ 
```

```python
coef_ 
```

```python
m = (coef_[0]/coef_[1])
```

```python
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize = (10,6))
plt.plot(x_input , y_input , color = "red", linewidth = 3 )
plt.scatter(X[:,0],X[:,1] , c = y ,cmap ="winter",s = 100 )
plt.ylim(-3,2)
```
