```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
x1 = [1,1,0,0]
x2 = [1,0,1,0]
```

```python
for df in [or_data,and_data,xor_data]:
  df['x1'] = x1
  df['x2'] = x2
```

```python
or_data["output"] = [a | b  for a , b in zip(x1,x2)]
and_data["output"] = [a & b  for a , b in zip(x1,x2)]
xor_data["output"] = [a^b  for a , b in zip(x1,x2)]
```

```python
or_data
```

```python
and_data
```

```python
xor_data
```

```python
plt.scatter(or_data["x1"],or_data["x2"],c =or_data["output"] ,cmap ="winter")
```

```python
plt.scatter(and_data["x1"],and_data["x2"],c =and_data["output"] ,cmap ="winter")
```

```python
plt.scatter(xor_data["x1"],xor_data["x2"],c =xor_data["output"] ,cmap ="winter")
```

```python
from sklearn.linear_model import Perceptron
clf1 =Perceptron()
clf2 =Perceptron()
clf3 =Perceptron()
```

```python
clf1.fit(or_data.iloc[:,0:2].values , or_data.iloc[:,2].values )
clf2.fit(and_data.iloc[:,0:2].values , and_data.iloc[:,2].values )
clf3.fit(xor_data.iloc[:,0:2].values , xor_data.iloc[:,2].values )
```

```python
clf1.coef_
```

```python
clf1.intercept_
```

```python
x = np.linspace(-1,1,5)
```

```python
y = -x+0.5
```

```python
plt.plot(x,y)
sns.scatterplot(data=or_data, x='x1', y='x2', hue='output', s=200)
plt.show()
```

```python
x = np.linspace(-1,1,5)
y = -x+1
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
# 2. Plotting the decision region
plt.figure(figsize=(8, 6))
# Using .values converts the DataFrame slice into the required NumPy array format
plot_decision_regions(X=xor_data.iloc[:, :2].values,
                      y=xor_data.iloc[:, -1].values.astype(int),
                      clf=clf3,
                      legend=2)

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('XOR Decision Region')
plt.show()
```
